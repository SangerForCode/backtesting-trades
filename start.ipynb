```python
import yfinance as yf
import pandas as pd

data = yf.download(
    "^NSEI",
    period="60d",
    interval="15m"  # Options: "1m", "5m", "15m", "1h", "1d", "1wk"
)

# Clean Nifty Data
data.dropna(inplace=True)
data = data[data['Close'] > 0]
data = data[data['Open'] > 0]
data.reset_index(inplace=True)
# # Ensure the index is datetime type for mplfinance
# data.index = pd.to_datetime(data.index)
data['Datetime'].head()
```

```python
rdata = pd.DataFrame(data.copy())
rdata.drop(['index','level_0'],inplace=False, axis=1)
rdata=rdata.set_index('Datetime',inplace=True)
```

```python
# Plot Nifty data using mplfinance with updated title
import mplfinance as mpf

mpf.plot(rdata, type='candle', style='charles', title='Nifty 50 Candlestick Pattern', ylabel='Price (INR)', volume=True)
```

# Candlestick Plot
Visualize the Nifty data in a candlestick pattern to analyze price movements.

```python
# Candlestick Plot
import plotly.graph_objects as go
 
fig = go.Figure(data=[go.Candlestick(x=data['Date'],
                open=data['Open'],
                high=data['High'],
                low=data['Low'],
                close=data['Close'])])
 
fig.update_layout(title='Nifty 50 Candlestick Chart',
                  xaxis_title='Date',
                  yaxis_title='Price (INR)',
                  xaxis_rangeslider_visible=False)
fig.show()
```
